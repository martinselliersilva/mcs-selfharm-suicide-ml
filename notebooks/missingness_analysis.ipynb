```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = "../data/mcs_sh_sample_Feb_28.csv"
missing_threshold = 50
output_path = "missingness_summary_mcs_sh_sample.csv"

df = pd.read_csv(file_path)
print("Dataset shape:", df.shape)

missing_summary = pd.DataFrame({
    "Variable": df.columns,
    "Missing_Count": df.isna().sum(),
    "Missing_Percent": df.isna().mean() * 100
}).sort_values("Missing_Percent", ascending=False).reset_index(drop=True)

missing_summary[f"High_Missing_over_{missing_threshold}_percent"] = (
    missing_summary["Missing_Percent"] > missing_threshold
)

print("\nTop 25 variables by missingness:\n")
print(missing_summary.head(25))

num_high = missing_summary[f"High_Missing_over_{missing_threshold}_percent"].sum()
print(f"\nVariables with over {missing_threshold}% missingness:", int(num_high))

plt.figure(figsize=(10, 6))
plt.hist(missing_summary["Missing_Percent"], bins=30)
plt.xlabel("Missing percent")
plt.ylabel("Number of variables")
plt.title("Missingness distribution in mcs sh sample")
plt.show()

missing_summary.to_csv(output_path, index=False)
print("\nSaved missingness summary to:", output_path)
```
