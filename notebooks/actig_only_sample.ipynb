```python
import pandas as pd
import numpy as np

in_path = "../data/mcs_sh_sample_Feb_28_preprocessed.csv"
out_path = "../data/mcs_sh_sample_actig_Feb_28_preprocessed.csv"

df = pd.read_csv(in_path)
df.columns = df.columns.str.strip()

# Normalize missing markers so the filter works
df = df.replace(["", " ", "NA", "N/A", "null", "None", "nan"], np.nan)

# Keep only rows where mean_acc_24h is present (ie participant has actigraphy)
df_act = df[df["mean_acc_24h"].notna()].copy()

print(f"Actigraphy only sample size: {df_act.shape[0]} out of {df.shape[0]}")

df_act.to_csv(out_path, index=False)
print(f"✅ Saved actigraphy only dataset to: {out_path}")
```
